```python
from sklearn.cluster import MiniBatchKMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
X, _=make_blobs(n_samples=1500,centers=3, n_features=3,random_state=42)
mbk_means=MiniBatchKMeans(n_clusters=4, batch_size=100, random_state=42)
mbk_means.fit(X)
plt.scatter(X[:,0],X[:,1],c=mbk_means.labels_, cmap='viridis')
plt.scatter(mbk_means.cluster_centers_[:,0],mbk_means.cluster_centers_[:,1],s=200,marker='X')
plt.title('Mini-Batch K-Means Clustering')
plt.show()
```
